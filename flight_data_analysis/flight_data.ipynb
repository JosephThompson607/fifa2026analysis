```python
def convert_to_east_coast_time(row):
    # Combine date and time to a naive datetime object
    naive_dt = pd.to_datetime(f"{row['date']} {row['time']}")

    # Extract offset in hours from the 'tz' string (e.g., 'UTC-6' -> -6)
    tz_offset_str = row['tz'].replace('UTC', '')
    offset_hours = int(tz_offset_str) if tz_offset_str else 0

    # Convert the naive local datetime to UTC datetime
    # If the local time is X hours behind UTC (e.g., UTC-6), then to get UTC, we add X hours to local time.
    # So, `utc_dt = naive_dt - pd.Timedelta(hours=offset_hours)` correctly adjusts.
    # Example: 13:00 UTC-6. offset_hours = -6. utc_dt = 13:00 - (-6) hours = 13:00 + 6 hours = 19:00 UTC.
    utc_dt = naive_dt - pd.Timedelta(hours=offset_hours)

    # Localize the UTC datetime to UTC and then convert to 'America/New_York' timezone
    aware_utc_dt = utc_dt.tz_localize('UTC')
    east_coast_dt = aware_utc_dt.tz_convert('America/New_York')
    return east_coast_dt

df = pd.DataFrame(matches)

country_order = ['Mexico', 'USA', 'Canada',]

df['country'] = pd.Categorical(
    df['country'],
    categories=country_order,
    ordered=True
)

stadium_order = (
    df[['country', 'lat', 'stadium']]
    .drop_duplicates()
    .sort_values(['country', 'lat'])['stadium']
    .tolist()
)
df['montreal_time'] = df.apply(convert_to_east_coast_time, axis=1)
#Artificial end time to get boxes on gannt chart
df['montreal_end_time'] = df['montreal_time'] + pd.Timedelta(hours=3)
```

```python
def haversine_vectorized(lat1, lon1, lat2, lon2):
    R = 6371.0  # Earth's radius in km
    # Convert degrees to radians
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c
#Get approximate flight distance
connections_df['flight_dist_km'] = haversine_vectorized(
    connections_df['SourceLatitude'], connections_df['SourceLongitude'],
    connections_df['DestinationLatitude'], connections_df['DestinationLongitude']
)
```

```python
from itertools import product

def airports_near_locations(airports_df, locations_df, max_distance_km,
                             airport_lat='Latitude', airport_lon='Longitude',
                             location_lat='camp_lat', location_lon='camp_long'):

    pairs = pd.DataFrame(
        list(product(airports_df.index, locations_df.index)),
        columns=['airport_idx', 'location_idx']
    )

    pairs['airport_lat'] = airports_df.loc[pairs['airport_idx'], airport_lat].values
    pairs['airport_lon'] = airports_df.loc[pairs['airport_idx'], airport_lon].values
    pairs['location_lat'] = locations_df.loc[pairs['location_idx'], location_lat].values
    pairs['location_lon'] = locations_df.loc[pairs['location_idx'], location_lon].values

    pairs['distance_km'] = haversine_vectorized(
        pairs['airport_lat'], pairs['airport_lon'],
        pairs['location_lat'], pairs['location_lon']
    )

    pairs = pairs[pairs['distance_km'] <= max_distance_km].reset_index(drop=True)

    result = pairs.merge(airports_df, left_on='airport_idx', right_index=True) \
                  .merge(locations_df, left_on='location_idx', right_index=True,
                         suffixes=('_airport', '_location'))

    return result.drop(columns=['airport_idx','location_idx', 'airport_lat', 'airport_lon',
                                 'location_lat', 'location_lon'])

```

```python
from itertools import permutations

def query_airport_pairs(G,  comparison_list):
    """
    For a list of airport IDs, checks all pairs for:
    - Direct flight existence
    - Number of direct flights
    - Distance
    - Shortest path with layovers
    """
    results = []

    for src, dst in comparison_list:
        if G.degree(src)==0:
            print("src has no flights")
            continue
        if G.degree(dst)==0:
            print("destination has no flights")
            continue
        if src==dst:
            continue
        result = {
            "SourceAirportID": src,
            "DestinationAirportID": dst,
            "has_direct_flight": False,
            "n_direct_flights": 0,
            "direct_distance_km": None,
            "has_any_path": False,
            "n_layovers": None,
            "layover_airports": []
        }

        # Check direct flight (edge in either direction)
        if G.has_edge(src, dst):
            edge = G[src][dst]
            result["has_direct_flight"] = True
            result["n_direct_flights"] = edge.get("n_flights", 0)
            result["direct_distance_km"] = edge.get("dist", None)

        # Shortest path (layovers) — works even if no direct flight
        try:
            path = nx.shortest_path(G, src, dst)
            result["has_any_path"] = True
            result["n_layovers"] = len(path) - 2  # subtract src and dst
            result["layover_airports"] = path[1:-1]  # intermediate nodes
            result["path_distance_km"] = sum(
                                    G[u][v].get("dist", 0) for u, v in zip(path, path[1:])
                                )
        except nx.NetworkXNoPath:
            pass
        except nx.NodeNotFound as e:
            print(f"Node not found: {e}")

        results.append(result)

    return results

#res = query_airport_pairs(G, permutations(id_list, 2) )
```
