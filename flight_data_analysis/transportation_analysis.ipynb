```python
matches = pd.DataFrame(matches)
matches['lat'] = matches['lat'].astype(float)
matches['lon'] = matches['lon'].astype(float)
# Melt so each row is one team's appearance in a game
team1_df = matches[['match', 'date', 'time', 'tz', 'team1', 'stadium', 'lat', 'lon', 'city_name', 'country']].copy()
team1_df = team1_df.rename(columns={'team1': 'team', 'team2': 'opponent'} if False else {'team1': 'team'})
team1_df['opponent'] = matches['team2']

team2_df = matches[['match', 'date', 'time', 'tz', 'team2', 'stadium','lat', 'lon', 'city_name', 'country']].copy()
team2_df = team2_df.rename(columns={'team2': 'team'})
team2_df['opponent'] = matches['team1']

long_df = pd.concat([team1_df, team2_df], ignore_index=True).sort_values(['team', 'date'])

# Number each team's games 1, 2, 3
long_df['game_num'] = long_df.groupby('team').cumcount() + 1

# Pivot wide — one row per team, columns for each game
result = long_df.pivot(index='team', columns='game_num',
                       values=['date', 'time', 'tz', 'stadium','lat', 'lon',  'city_name', 'country', 'opponent'])


# Flatten column names: date_1, date_2, ...
result.columns = [f'{col}_{num}' for col, num in result.columns]
result = result.reset_index()
for col in ['lat_1', 'lon_1', 'lat_2', 'lon_2', 'lat_3', 'lon_3']:
    result[col] = result[col].astype(float)
```

```python
def haversine_vectorized(lat1, lon1, lat2, lon2):
    R = 6371.0  # Earth's radius in km
    # Convert degrees to radians
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c
```

```python

flights_df = pd.read_csv('stadium_flight_distances.csv')

flight_lookup = flights_df.set_index(['origin_stadium', 'destination_stadium'])

def lookup_flight(row, from_col, to_col):
    origin = row[from_col]
    dest = row[to_col]

    r = flight_lookup.loc[(origin, dest)]
    
    return r['total_direct_flights'],  r['avg_direct_distance_km']


# Leg 1→2
result[['flights_1_2', 'flight_distance_1_2']] = result.apply(
    lambda row: pd.Series(lookup_flight(row, 'stadium_1', 'stadium_2')) 
    if row['stadium_1'] != row['stadium_2'] else pd.Series([0, 0]), axis=1
)

# Leg 2→3
result[['flights_2_3', 'flight_distance_2_3']] = result.apply(
    lambda row: pd.Series(lookup_flight(row, 'stadium_2', 'stadium_3')) 
    if row['stadium_2'] != row['stadium_3'] else pd.Series([0,  0]), axis=1
)

result['abs_dist_1_2'] = haversine_vectorized(result['lat_1'],result['lon_1'],result['lat_2'],result['lon_2'])
result['abs_dist_2_3'] = haversine_vectorized(result['lat_2'],result['lon_2'],result['lat_3'],result['lon_3'])


result
```
