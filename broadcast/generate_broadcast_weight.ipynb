```python
# Calculate per-capita metrics
df["wiki_window_views_per_capita"] = df["wiki_window_views"] / df["population"]
df["combined_per_capita"] = df["combined"] / df["population"]
```

```python
df.sort_values('wiki_window_views_per_capita', inplace=True, ascending=False)
df[['country', 'population', 'wiki_window_views', 'wiki_window_views_per_capita', 'combined', 'combined_per_capita']].head(10)
```

```python
df.sort_values('combined_per_capita', inplace=True, ascending=False)
df[['country', 'population', 'wiki_window_views', 'wiki_window_views_per_capita', 'combined', 'combined_per_capita']].head(10)
```

```python
df["trends_interest_per_capita"] = df["trends_interest"] * df["population"]
```

```python
df.sort_values('trends_interest_per_capita', inplace=True, ascending=False)
df[['country', 'population', 'trends_interest_per_capita', 'wiki_window_views_per_capita', 'combined', 'combined_per_capita']].head(len(df))
```

```python
df_odds = pd.read_csv('betting_team_probabilities.csv')
```

```python
# Merge df_odds with df to get trends_interest_per_capita
df_odds = df_odds.merge(df[['country', 'trends_interest_per_capita']], on='country', how='left')

# Create game_weight_for_country
df_odds['game_weight_for_country'] = df_odds['closeness'] * df_odds['trends_interest_per_capita']

df_odds[['country', 'closeness', 'trends_interest_per_capita', 'game_weight_for_country']].head(10)
```

```python
df_odds.sort_values('game_weight_for_country', inplace=True, ascending=False)
df_odds[['country', 'closeness', 'trends_interest_per_capita', 'game_weight_for_country']].head(len(df_odds))
```
